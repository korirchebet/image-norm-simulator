```python
#normalization using min-max scaling example
import numpy as np
data =np.array([1,2,3,4,5],dtype = float)
min_value = np.min(data)
max_value = np.max(data)
scaled_data = (data - min_value)/ (max_value - min_value)
scaled_data
```

```python
#normalization using z-score normalization
data = np.array([5,6,7,8,9],dtype=float)
mean_val=np.mean(data)
std_val = np.std(data)
norm_data= (data - mean_val) / std_val
norm_data
```

```python
#normalization by the L2 technique
data = np.array([1,2,3,4,5],dtype = float)
magnitude = np.linalg.norm(data)
norm_data = data/magnitude
print("Original data :",data)
print("Normalized data:",norm_data)
```

```python
#image normalization
image = np.array([
    [[10,20,30],[40,50,60],[70,80,90]],
    [[15,25,35],[45,55,65],[75,85,95]]
])
norm_image = image/255.0
#calculating the mean pixel intensity per channel and normalizing each channel to range[0,1]
mean_per_channel= np.mean(image,axis =(0,1))
print("Image shape:",image.shape)
print("Normalized Image:",norm_image)
print("Mean Per Channel:",mean_per_channel)
```

```python
#computing using loops
#convert type to float
image = image.astype(np.float32)
#create a copy for looping
norm_loop= image.copy()
#for loop
for c in range(image.shape[2]):
    channel = image[:,:,c]
    mean = np.mean(channel)
    std = np.std(channel)
    norm_loop[:,:,c]=(channel - mean)/std
print("Original Image:\n",image)
print("Loop:\n",norm_loop)
```

```python
#computing using broadcasting
mean = np.mean(image,axis=(0,1), keepdims=True)
std= np.std(image,axis=(0,1),keepdims = True)
broad_data = (image-mean)/std
print("Broadcasted Normalization:\n",broad_data)
               
```

```python
#compare the speed between using loops and broadcsting   
# Measure loop version speed
%timeit for c in range(image.shape[2]): (image[:, :, c] - np.mean(image[:, :, c])) / np.std(image[:, :, c])

# Measure broadcasting version speed
%timeit (image - np.mean(image, axis=(0, 1), keepdims=True)) / np.std(image, axis=(0, 1), keepdims=True)
```
